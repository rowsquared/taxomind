```python
import pandas as pd
import uuid
from pathlib import Path
import json
```

```python
RAW_DATA_DIR = Path("../data/01_raw")
```

```python
df_processed = pd.read_csv(RAW_DATA_DIR/"isco_train_data.csv")
df_processed.head()
```

```python
sentences_dict = {"taxonomyKey": "ISCO", "sentences": []}
for _, row in df_processed.iterrows():
    sentence_id = row.get("id") or str(uuid.uuid4())
    fields = {
        "Job Description": row["Job Description"],
        "Industry Description": row["Industry Description"],
    }
    snetence = {"sentence_id": sentence_id, "fields": fields}
    sentences_dict["sentences"].append(snetence)
```

```python
with open(RAW_DATA_DIR/"isco_test_sentences.json", "w") as f:
    json.dump(sentences_dict, f, indent=2)
```
